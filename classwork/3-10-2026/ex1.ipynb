```python
df = pd.DataFrame(
    {
        "sales": [200.0, 320.0, 150.0, 480.0, 220.0, 390.0, 180.0, 510.0],
        "ad_camp": [10.0, 25.0, 5.0, 40.0, 12.0, 30.0, 8.0, 45.0],
        "campaign": [
            "Email",
            "Social",
            "Email",
            "TV",
            "Email",
            "Social",
            "Email",
            "TV",
        ],
    }
)
```

```python
import patsy
```

```python
y, X = patsy.dmatrices("sales ~ ad_camp + campaign", data=df, return_type="dataframe")
print(X.shape)
print(X)
print(X.design_info)
```

```python
from statsmodels.api import OLS
from statsmodels.formula.api import ols
```

```python
ols_model = ols("sales ~ ad_camp + campaign", data=df).fit()
ols_model.summary()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
x_train = patsy.dmatrix("ad_camp + campaign", data=df, return_type="dataframe")
y_train = df["sales"].to_numpy()
```

```python
reg = LinearRegression(fit_intercept=False)
reg.fit(x_train, y_train)
```
